1. Яка кількість людей в кожній філії робили покупку як Member чи Normal

```python
df.pivot_table(index='Branch', columns='Customer type', values='Invoice ID', aggfunc='count')
```

У трьох філіях приблизно половина покупок робиться клієнтами Member, а друга половина звичайними клієнтами ,Normal. Бізнесу варто запустити акції або пропозиції для коверції покупців з Normal до Member

2. На яку загальну суму зробили покупки покупці типу Member та Normal за філіями

```python
df.pivot_table(index='Branch', columns='Customer type', values='Total', aggfunc='sum').round(2)
```

За всіма філіями Members зробили покупки на більшу загальну суму ніж Normal покупці. Бізнес може запровадити персоналізовані акції та діскаунти для Members щоби заохочувати покупців бути Members та current members робити більше покупок за дешевшими цінами з акціями

3. Яка категорія товарів приносить найбільший прибуток

```python
df.groupby('Product line')['gross income'].sum().round(2)
```

Найбільший прибуток приносить категорія Food and beverages, тому треба завжди підтримувати наявність цих товарів.
Найменший прибуток приносить категорія Health and beauty, тому треба запровадити акції або змінити розміщення товару для цієї категорії. Різниця між всіма категоріями невелика, тож прибуток розподілений досить рівномірно.
